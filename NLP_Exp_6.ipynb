```python
import nltk
from nltk.corpus import brown
from nltk.tag import hmm
```

```python
nltk.download('brown')
nltk.download('universal_tagset')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
data = brown.tagged_sents(
    tagset='universal'
)

print(data[:2])
```

```python
train_data = data[:3000]
test_data = data[3000:3100]

print("Training:", len(train_data))
print("Testing:", len(test_data))
```

```python
trainer = hmm.HiddenMarkovModelTrainer()

tagger = trainer.train(train_data)

print("HMM Model Created")
```

```python
sentence = [
    "Formula", "one", "drivers", "race", "fast"
]

print(sentence)
```

```python
result = nltk.pos_tag(sentence, tagset='universal')

print("POS Tags:")
print(result)
```

```python
for word, tag in result:
    print(word, "->", tag)
```

```python
sentence2 = [
    "The", "driver", "wins", "the", "race"
]

print(tagger.tag(sentence2))
```

```python
best_path = tagger.tag(sentence2)

print("Best Viterbi Path:")
print(best_path)
```

```python
import warnings

with warnings.catch_warnings():
    warnings.simplefilter("ignore", RuntimeWarning)
    accuracy = tagger.accuracy(test_data)

print("HMM Accuracy:", accuracy)
```

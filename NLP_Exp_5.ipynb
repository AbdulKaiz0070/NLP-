```python
import nltk
from nltk.util import ngrams
from collections import Counter
from nltk.metrics import edit_distance
import math
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = """Formula one racing is fast.
Racing drivers compete in different circuits.
Teams use fast racing cars and tyres.
Drivers compete for points in every race."""
```

```python
words = nltk.word_tokenize(text.lower())
print(words)
```

```python
unigrams = list(ngrams(words, 1))
print("Unigrams:", unigrams)
```

```python
bigrams = list(ngrams(words, 2))
print("Bigrams:", bigrams)
```

```python
trigrams = list(ngrams(words, 3))
print("Trigrams:", trigrams)
```

```python
bigram_count = Counter(bigrams)

print("Most Common Bigrams:")
print(bigram_count.most_common(10))
```

```python
def bigram_probability(w1, w2):
    return bigram_count[(w1, w2)] / words.count(w1)

print("P(racing | formula):",
      bigram_probability("formula", "racing"))
```

```python
sentence = ["formula", "one", "racing"]
pairs = list(ngrams(sentence, 2))

prob = math.prod(
    bigram_probability(a, b) for a, b in pairs
)

print("Sentence Probability:", prob)
```

```python
print("Edit Distance:")
print("racing → race =", edit_distance("racing", "race"))
print("driver → drivers =", edit_distance("driver", "drivers"))
```

```python
N = len(pairs)

perplexity = (1 / prob) ** (1 / N)

print("Test Sentence:", "formula one racing")
print("Perplexity:", perplexity)
```

```python

import gensim.downloader as api
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

print("Dependencies loaded successfully.")
```

```python
print("Loading word embeddings (GloVe 100d)...")
word_vectors = api.load("glove-wiki-gigaword-100")
print("Embeddings loaded successfully!")
```

```python
def expand_query(query: str, top_k: int = 2, similarity_threshold: float = 0.6) -> str:

    tokens = word_tokenize(query.lower())
    expanded_terms = list(tokens)

    for token in tokens:
        if token in word_vectors:
            # Retrieve top similar words in vector space
            similar_words = word_vectors.most_similar(token, topn=top_k)
            for word, score in similar_words:
                if score >= similarity_threshold and word not in expanded_terms:
                    expanded_terms.append(word)

    return " ".join(expanded_terms)

print("Query expansion logic defined.")
```

```python
queries = [
    "ferrari race speed",
    "formula 1 engine power",
    "automobile maintenance tips"
]

print(f"{'Original Query':<30} | {'Expanded Query'}")
print("-" * 80)

for q in queries:
    expanded = expand_query(q, top_k=2, similarity_threshold=0.55)
    print(f"{q:<30} | {expanded}")
```

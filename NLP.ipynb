```python
text = input("Enter text: ")

tokens = word_tokenize(text.lower())

words = [word for word in tokens if word.isalpha()]

fd = FreqDist(words)

print(fd)

print("\nWord Frequencies")

for word, count in fd.items():
    print(word, ":", count)
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from collections import Counter
```

```python
def perform_nlp(text, language):

    print("\n====================")
    print(language)
    print("====================")

    # Sentence count
    sentences = sent_tokenize(text)
    print("\nSentence Count:")
    print(len(sentences))


    # Word count
    words = word_tokenize(text)
    print("\nWord Count:")
    print(len(words))


    # Tokens
    print("\nTokens:")
    print(words)


    # Uppercase and Lowercase only for languages that support it
    case_languages = [
        "English",
        "French",
        "German",
        "Spanish"
    ]

    if language in case_languages:

        print("\nLowercase:")
        print(text.lower())

        print("\nUppercase:")
        print(text.upper())

    else:

        print("\nLowercase:")
        print("Not applicable (language has no uppercase/lowercase system)")

        print("\nUppercase:")
        print("Not applicable (language has no uppercase/lowercase system)")


    # Reverse text
    print("\nReverse Text:")
    print(text[::-1])


    # Reverse words
    print("\nReverse Words:")
    print(" ".join(words[::-1]))


    # Word frequency
    print("\nWord Frequency:")
    print(Counter(words))
```

```python
perform_nlp(
    texts["French"],
    "French"
)
```

```python
perform_nlp(
    texts["Japanese"],
    "Japanese"
)
```

```python
tamil_nlp = stanza.Pipeline("ta")
```

```python
tamil_text = "நான் தமிழ் மொழியை விரும்புகிறேன்."

print("Tamil Text:")
print(tamil_text)
```

```python
doc = tamil_nlp(tamil_text)

tamil_words = []

for sentence in doc.sentences:
    for word in sentence.words:
        tamil_words.append(word.text)

print("Tamil Tokens:")
print(tamil_words)

print("\nWord Count:")
print(len(tamil_words))
```

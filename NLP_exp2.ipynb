```python
import nltk
import re

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

print("Libraries imported successfully!")
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = """
Max Verstappen is racing quickly at the Monaco Grand Prix!
The Red Bull Racing team is changing tyres during the pit stop.
Lewis Hamilton is driving a fast Mercedes car, and Ferrari is preparing for qualifying.
Drivers are competing for championship points, while engineers are improving race strategies.
"""

print(text)
```

```python
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)
```

```python
lowercase_text = text.lower()

print("After Case Folding:")
print(lowercase_text)
```

```python
clean_text = re.sub(r'[^\w\s]', '', lowercase_text)

print("After Punctuation Removal:")
print(clean_text)
```

```python
tokens = word_tokenize(clean_text)

print("Clean Tokens:")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print("After Stop-word Removal:")
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()

print("Stemming:")

for word in filtered_tokens:
    print(word, "->", stemmer.stem(word))
```

```python
# Cell 10 - Automatic Lemmatization

from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

lemmatizer = WordNetLemmatizer()

f1_words = [
    "racing",
    "driving",
    "changing",
    "running",
    "winning",
    "competing",
    "overtaking",
    "drivers",
    "teams",
    "circuits"
]

print("F1 Lemmatization:\n")

for word in f1_words:
    if word.endswith("ing"):
        lemma = lemmatizer.lemmatize(word, wordnet.VERB)
    else:
        lemma = lemmatizer.lemmatize(word, wordnet.NOUN)

    print(word, "->", lemma)
```

```python
# Step 1: Case folding
normalized_text = text.lower()

# Step 2: Remove punctuation
normalized_text = re.sub(r'[^\w\s]', '', normalized_text)

# Step 3: Tokenization
tokens = word_tokenize(normalized_text)

# Step 4: Stop-word removal
tokens = [
    word for word in tokens
    if word not in stop_words
]

# Step 5: Lemmatization
normalized_tokens = [
    lemmatizer.lemmatize(word)
    for word in tokens
]

print("Final Normalized Text:")
print(normalized_tokens)
```

```python
print("=" * 50)
print("FORMULA 1 TEXT NORMALIZATION")
print("=" * 50)

print("Original Text:")
print(text)

print("\nNormalized Tokens:")
print(normalized_tokens)

print("=" * 50)
```
